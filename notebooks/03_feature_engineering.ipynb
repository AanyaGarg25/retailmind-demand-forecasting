```python
import pandas as pd 
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/retail_store_inventory.csv")

df["Date"] = pd.to_datetime(df["Date"])

df.head()
```

```python
df = df.sort_values(
    by=["Store ID", "Product ID", "Date"]
)

df.head()
```

```python
df["Year"] = df["Date"].dt.year

df["Month"] = df["Date"].dt.month

df["Quarter"] = df["Date"].dt.quarter

df["Day"] = df["Date"].dt.day

df["DayOfWeek"] = df["Date"].dt.dayofweek

df["DayOfYear"] = df["Date"].dt.dayofyear

df["WeekOfYear"] = df["Date"].dt.isocalendar().week.astype(int)

df.head()
```

```python
df["IsWeekend"] = np.where(
    df["DayOfWeek"] >= 5,
    1,
    0
)

df[["Date", "DayOfWeek", "IsWeekend"]].head()
```

```python
df["lag_7"] = (
    df.groupby(
        ["Store ID", "Product ID"]
    )["Units Sold"]
    .shift(7)
)
```

```python
yearly_sales = (
    df.set_index("Date")
    .resample("YE")["Units Sold"]
    .sum()
)

print(yearly_sales)
```

```python
df["lag_14"] = (
    df.groupby(
        ["Store ID", "Product ID"]
    )["Units Sold"]
    .shift(14)
)
```

```python
df["lag_30"] = (
    df.groupby(
        ["Store ID", "Product ID"]
    )["Units Sold"]
    .shift(30)
)
```

```python
df["lag_90"] = (
    df.groupby(
        ["Store ID", "Product ID"]
    )["Units Sold"]
    .shift(90)
)
```

```python
df[
    [
        "Date",
        "Store ID",
        "Product ID",
        "Units Sold",
        "lag_7",
        "lag_14"
    ]
].head(20)
```

```python
df["rolling_mean_7"] = (
    df.groupby(
        ["Store ID", "Product ID"]
    )["Units Sold"]
    .transform(
        lambda x: x.shift(1).rolling(7).mean()
    )
)
```

```python
df["rolling_mean_30"] = (
    df.groupby(
        ["Store ID", "Product ID"]
    )["Units Sold"]
    .transform(
        lambda x: x.shift(1).rolling(30).mean()
    )
)
```

```python
df["rolling_std_7"] = (
    df.groupby(
        ["Store ID", "Product ID"]
    )["Units Sold"]
    .transform(
        lambda x: x.shift(1).rolling(7).std()
    )
)
```

```python
df["rolling_std_30"] = (
    df.groupby(
        ["Store ID", "Product ID"]
    )["Units Sold"]
    .transform(
        lambda x: x.shift(1).rolling(30).std()
    )
)
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=["lag_90"])

df = df.dropna()

df.shape
```
